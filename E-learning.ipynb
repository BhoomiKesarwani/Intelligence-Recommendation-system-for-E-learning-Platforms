```python
# Phase 12: Content-Based Recommendation System

import pandas as pd
courses = pd.read_csv(r"C:\Users\krish\OneDrive\Desktop\IRS for E-Learning Platforms/clean_courses.csv")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
courses["combined_features"].isnull().sum()
```

```python
courses["combined_features"] = courses["combined_features"].fillna("")
```

```python
# A TF-IDF Vector is a numerical representation of text that tells a machine how important each word is in a document.

# 1: Term Frequency (TF)
# TF measures how often a word appears in a document.
# Example:
# Python Python Data Analysis
# Word	Count
# Python	2
# Data	1
# Analysis 1

# 2: Inverse Document Frequency (IDF)
# Some words appear in almost every document, such as:
# course
# learn
# introduction
# These words are not very useful for distinguishing documents.
# IDF gives lower importance to common words and higher importance to rare, informative words.
# Example:
# Word	Appears in	Importance
# Python	50 courses	Medium
# Machine Learning	10 courses	High
# Course	500 courses	Very Low
```

```python
# Convert Text into TF-IDF Vectors
tfidf = TfidfVectorizer(stop_words="english")
# tfidf_matrix = tfidf.fit_transform(courses["combined_features"])
```

```python
tfidf_matrix = tfidf.fit_transform(courses["combined_features"])
```

```python
tfidf_matrix.shape
```

```python
# cosine_similarity():::This is a function from Scikit-learn that compares two vectors and returns a similarity score.
# score ranges from 0-1

cosine_sim=cosine_similarity(tfidf_matrix)
cosine_sim.shape
```

```python
# Create an Index Mapping
indices = pd.Series(
    courses.index,
    index=courses["course_name"]
).drop_duplicates()
```

```python
indices.head()
```

```python
# Build the Recommendation Function

def recommend_courses(course_name, top_n=10):

    if course_name not in indices:
        return "Course not found."

    idx = indices[course_name]

    similarity_scores = list(enumerate(cosine_sim[idx]))

    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )
    

    similarity_scores = similarity_scores[1:top_n+1]

    course_indices = [i[0] for i in similarity_scores]

    return courses[
        [
            "course_name",
            "category",
            "platform",
            "rating",
            "level"
        ]
    ].iloc[course_indices]
                             
    
```

```python
courses["course_name"].sample(10)  #iloc = Integer Location -->It selects rows and columns using row numbers, not column values.
     #enumerate() adds the position (index) to each value.
```

```python
# Phase 13: Collaborative Filtering-->Instead of looking at course descriptions, collaborative filtering recommends courses based on user behavior.
# import pandas as pd

# interactions = pd.read_csv(r"C:\Users\krish\OneDrive\Desktop\IRS for E-Learning Platforms/research_interactions.csv")
# courses = pd.read_csv(r"C:\Users\krish\OneDrive\Desktop\IRS for E-Learning Platforms/clean_courses.csv")
```

```python
interactions.head()
```

```python
# Create the User-Course Matrix
user_course_matrix = interactions.pivot_table(
    index="user_id",
    columns="course_id",
    values="rating"
)
```

```python
user_course_matrix.shape
```

```python
user_course_matrix = user_course_matrix.fillna(0) #Fill Missing Ratings
```

```python
user_similarity = cosine_similarity(user_course_matrix)

# Convert it into a DataFrame for easier access
user_similarity_df = pd.DataFrame(
    user_similarity,
    index=user_course_matrix.index,
    columns=user_course_matrix.index
)
```

```python
# Find Similar Users
user_similarity_df.loc["U00001"].sort_values(ascending=False).head(10)
```

```python
# Build a Collaborative Recommendation Function

def collaborative_recommend(user_id, top_n=10):

    # Similar users
    similar_users = (
        user_similarity_df[user_id]
        .sort_values(ascending=False)
        .index[1:11]
    )

    # Courses already rated
    watched = interactions[
        interactions["user_id"] == user_id
    ]["course_id"].tolist()

    # Ratings from similar users
    recommendations = interactions[
        interactions["user_id"].isin(similar_users)
    ]

    recommendations = (
        recommendations.groupby("course_id")["rating"]
        .mean()
        .sort_values(ascending=False)
    )

    recommendations = recommendations[
        ~recommendations.index.isin(watched)
    ]

    return recommendations.head(top_n)
```
